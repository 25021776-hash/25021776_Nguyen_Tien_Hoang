```python
#W4A6
from math import *
def nto(c):
    if c < 2 :
        return False
    elif c == 2 or c == 3:
        return True
    else:
        for i in range (2,isqrt(c) + 1):
            if c % i == 0:
                return False
        return True
a,b = map(int,input().split())
for i in range(a,b+1):
    if nto(i):
        print(i,end=" ")
```

```python
#W4A7
from math import *
def nto(c):
    if c == 2 or c == 3:
        return True
    else:
        for i in range (2,isqrt(c) + 1):
            if c % i == 0:
                return False
        return True
n = int(input())
lon = 0
for i in range(2,n + 1):
    if nto(i):
        if n % i == 0:
            lon = i
print(lon)
```

```python
#W4A8
def reverse(n): # Ham lat nguoc so
    tem = 0
    while n != 0:
        tem = tem * 10 + n % 10
        n = n // 10
    return tem
def nguoc(n): #Ham ktra xem so doi xung
    return n == reverse(n)
if __name__ == "__main__":
    n = int(input())
    cnt = 0
    while True :
        if nguoc(n):
            print(cnt)
            print(n)
            break
        else:
            cnt += 1
            n = n + reverse(n)
```

```python
#W4A13
a, b = map(int,input().split())
dem1 = 0
dem2 = 0
for i in range(1,a):
    if a % i == 0:
        dem1 += i
for j in range(1,b):
    if b % j == 0:
        dem2 += j
if dem1 == b and dem2 == a:
    print("True")
else:
    print("False")
```
