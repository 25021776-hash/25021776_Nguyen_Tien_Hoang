```python
#W9A2
n,m = map(int, input().split())
k = int(input())
a = []
for i in range(n):
    a.append(list(map(int, input().split())))
def count(x):
    cnt = 0
    cot = 0
    hang = n-1
    while hang >= 0 and cot < m:
        if a[hang][cot] <= x:
            cnt += hang + 1
            cot += 1
        else:
            hang -= 1
    return cnt
l,r = a[0][0],a[n-1][m-1]
ans = 0
while l <= r:
    mid = (l + r) // 2
    if count(mid) >= k:
        ans = mid
        r = mid - 1
    else:
        l = m + 1
print(ans)


```

```python
#W9A3
s = list(input().strip())
n = len(s)
for i in range(n):
    minvalue = s[i]
    index1 = -9
    for j in range(n-1,i,-1):
        if s[j] < minvalue:
            minvalue = s[j]
            index1 = j
    if index1 != -9:
        s[i],s[index1] = s[index1],s[i]
        break
ans = "".join(s).lstrip("0")
print(ans)
```

```python
#W9A5
def compare(a,b):
    for i in range(len(b) - len(a) + 1):
        if b [i: i + len(a)] == a:
            return 'YES'
    return 'NO'
a = list(map(int, input().split()))
b = list(map(int, input().split()))
print(compare(a,b))
```

```python
#W9A7
def flower(a,k):
    n = len(a)
    if n == 1 and a[0] == 0:
        k -= 1
        return k <= 0
    if a[0] == 0 and a[1] == 0:
        k -= 1
        a[0] = 1
    for i in range(1, n-1):
        if a[i] == 0 and a[i-1] == 0 and a[i+1] == 0:
            k -= 1
            a[i] = 1
    if a[n - 1] == 0 and a[n - 2] == 0:
        k -= 1
        a[n - 1] = 1
    return k <= 0
a = list(map(int, input().split()))
k = int(input())
print(flower(a,k))
```
