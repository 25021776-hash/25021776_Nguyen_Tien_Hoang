```python
#W7A16
a = list(map(int, input().split()))
d = {}
for i in a:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1
check = 0
dem = 0
for k, v in d.items():
    if v > 0 and k == v:
        check = 1
    else:
        check = 0
        if v > k:
            dem += v - k
        else:
            dem += k
if check == 1:
    print("YES")
else:
    print("NO",dem)



```

```python
#W7A19
a = list(map(int, input().split()))
b = list(map(int, input().split()))
c = []
i,j = 0,0
while i < (len(a)) and j < len((b)):
    if a[i] < b[j]:
        c.append(a[i])
        i += 1
    else:
        c.append(b[j])
        j += 1
c.extend(a[i::])
c.extend(b[j::])
print(c)
```

```python
#W7A21
def tong(s):
    a = [31,29,31,30,31,30,31,31,30,31,30,31]
    if int(s[1]) < 1 or int(s[1]) > 12:
        return "INVALID"
    if int(s[0]) < 1 or int(s[0]) > 31:
        return "INVALID"
    sothutu = sum(a[0 :int(s[1]) - 1: 1 ]) + int(s[0])
    return sothutu

s = input()
s = s.split()
print(tong(s))
```
